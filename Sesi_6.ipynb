#MATERI

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-poster')

f = lambda t,s : np.exp(-t)
h = 0.1
t_min = 0
t_max = 1

t = np.arange(t_min,t_max+h,h)
s0 = -1

s = np.zeros(len(t))
s[0] = s0

for i in range(0,len(t)-1):
    s[i+1] = s[i] + h*f(t[i],s[i])

plt.figure(figsize = (12,8))
plt.plot(t,s, 'bo--', label = 'approx(euler h=0.1)')
plt.plot(t, -np.exp(-t), 'g', label = 'exact solution')
```

#MODUL

1. y' + 4y = x^2, y(0) = 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

f = lambda x,y: x**2 - 4*y
h = 0.1
x_min = 0
x_max = 1

x = np.arange(x_min, x_max + h, h)
y0 = 1

y = np.zeros(len(x))
y[0] = y0

for i in range(0, len(x) - 1):
  y[i+1] = y[i] + h*f(x[i], y[i])

plt.figure(figsize=(12, 8))
plt.plot(x, y, 'bo--', label = 'approx(euler h=0.01)')

exact_solution = lambda x: (x**2/4 - x/8 + 1/32) + (31/32)*np.exp(-4*x)

plt.plot(x, exact_solution(x), 'g', label = 'exact solution')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Solution of $y\' + 4y = x^2$ using Euler method')
plt.legend()
plt.grid(True)
plt.show()
```

2. y' = sin y, y(0) = 1

```python
from scipy.integrate import solve_ivp
f = lambda t,y: np.sin(y)
h = 0.1
t_min = 0
t_max = 5

t = np.arange(t_min, t_max + h, h)
y0 = 1

y = np.zeros(len(t))
y[0] = y0

for i in range(0, len(t) - 1):
  y[i+1] = y[i] + h*f(t[i], y[i])

sol = solve_ivp(f, [t_min, t_max], [y0], t_eval=t)

plt.figure(figsize=(12, 8))
plt.plot(t, y, 'bo--', label='Euler approx (h=0.1)')
plt.plot(t, sol.y[0], 'g', label='solve_ivp (accurate numerical solution)')

plt.xlabel('t')
plt.ylabel('y')
plt.title("Solusi dari $y' = \sin y$ dengan Metode Euler vs solve_ivp")
plt.legend()
plt.grid(True)
plt.show()
```
